```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('bike_rentals.csv')
```

```python
# total revenue by city
revenue = df.groupby('city')['rental_cost'].sum().sort_values(ascending=False)

plt.figure(figsize=(9, 5))
plt.bar(revenue.index, revenue.values, color='darkblue')

plt.title('Total Revenue by City')
plt.xlabel('City')
plt.ylabel('Total Revenue')
plt.xticks(rotation=45)
plt.show()
```

```python
# average  rating by city
avg_rating = df.groupby('city')['rating'].mean()

plt.figure(figsize=(9, 5))
plt.bar(avg_rating.index, avg_rating.values, color='darkred')

plt.title('Average Customer Rating by City')
plt.xlabel('City')
plt.ylabel('Average Rating')
plt.xticks(rotation=45)
plt.ylim(0, 5)
plt.show()
```

```python
# weeekdays vs weekends 
weekend_count = df['is_weekend'].value_counts()

plt.figure(figsize=(6, 5))
plt.bar(['Weekday', 'Weekend'], weekend_count.values, color=['gray', 'orange'])

plt.title('Rentals: Weekend vs Weekday')
plt.ylabel('Number of Rentals')
plt.show()
```

```python
# average cost per city
avg_cost = df.groupby('city')['rental_cost'].mean()

plt.figure(figsize=(9, 5))
plt.bar(avg_cost.index, avg_cost.values, color='navy')

plt.title('Average Rental Cost by City')
plt.xlabel('City')
plt.ylabel('Average Cost')
plt.xticks(rotation=45)
plt.show()
```

```python
# age distribution
plt.figure(figsize=(8, 5))
plt.hist(df['age'], bins=15, color='salmon', edgecolor='black')

plt.title('Age of Customers')
plt.xlabel('Age')
plt.ylabel('Number of People')
plt.grid(True)
plt.show()
```

```python
# distance vs duration
plt.figure(figsize=(8, 6))
plt.scatter(df['duration_hours'], df['distance_km'], alpha=0.5, color='blue')

plt.title('Duration vs Distance')
plt.xlabel('Duration (Hours)')
plt.ylabel('Distance (km)')
plt.grid(True)
plt.show()
```

```python
# boxplot of rental cost 
plt.figure(figsize=(8, 5))
plt.boxplot(df['rental_cost'])
plt.title('Rental Cost Distribution')
plt.ylabel('Cost')
plt.show()
```

```python
# membership type
mem = df['membership_type'].value_counts()

plt.figure(figsize=(8, 5))
plt.bar(mem.index, mem.values, color='green')

plt.title('Rentals by Membership Type')
plt.xlabel('Membership')
plt.ylabel('Count')
plt.xticks(rotation=30)
plt.show()
```

```python
# rating
rating_count = df['rating'].value_counts().sort_index()

plt.figure(figsize=(7, 5))
plt.bar(rating_count.index, rating_count.values, color='gold', edgecolor='black')

plt.title('Customer Ratings')
plt.xlabel('Rating (1 to 5)')
plt.ylabel('Number of Rentals')
plt.xticks([1,2,3,4,5])
plt.show()
```

```python
# weather condtions bar chart
weather_count = df['weather'].value_counts()

plt.figure(figsize=(8, 5))
plt.bar(weather_count.index, weather_count.values, color='orange')

plt.title('Rentals in Different Weather')
plt.xlabel('Weather')
plt.ylabel('Number of Rentals')
plt.show()
```

```python
    # pie chart  pyment method
payment = df['payment_method'].value_counts()

plt.figure(figsize=(7, 7))
plt.pie(payment.values, labels=payment.index, autopct='%1.1f%%', colors=['#ff9999','#66b3ff','#99ff99','#ffcc99'])
plt.title('How People Pay')
plt.show()
```

```python
# rental by city
city_counts = df['city'].value_counts()

plt.figure(figsize=(9, 5))
plt.bar(city_counts.index, city_counts.values, color='purple')

plt.title('Rentals in Each City')
plt.xlabel('City')
plt.ylabel('Number of Rentals')
plt.xticks(rotation=45)
plt.show()
```

```python
# bike type
bike_counts = df['bike_type'].value_counts()

plt.figure(figsize=(8, 5))
plt.bar(bike_counts.index, bike_counts.values, color='teal')

plt.title('Which Bike Type is Most Popular?')
plt.xlabel('Bike Type')
plt.ylabel('Number of Rentals')
plt.xticks(rotation=45)
plt.show()
```

```python
# rental distribution 
plt.figure(figsize=(8, 5))
plt.hist(df['duration_hours'], bins=15, color='skyblue', edgecolor='black')

plt.title('How Long People Rent Bikes')
plt.xlabel('Duration (Hours)')
plt.ylabel('Number of Rentals')
plt.grid(True)
plt.show()
```
