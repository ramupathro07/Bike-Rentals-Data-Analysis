```python
# What is the combined frequency of bike_type AND together?
df.groupby(['bike_type','city']).size().sort_values(ascending= False)
```

```python
# What is the average rental cost per city?
df.groupby('city')['rental_cost'].mean().round(2)
```

```python
# What is the min and max rental cost per bike type?
df.groupby('bike_type')['rental_cost'].agg(['max','min'])
```

```python
# What is the average age of renters per gender?
df.columns
df.groupby('gender')['age'].mean()
```

```python
# what is the total revenue collected per payment method?
df.groupby('payment_method')['rental_cost'].sum()
```

```python
# What is the average distance per membership type?
df.groupby('membership_type')['distance_km'].mean()
```

```python
# How many discounted rentals happened per city?
df[df['discount_applied']].groupby('city').size()
```

```python
# What is the average rental cost per city AND bike type?
df.groupby(['city','bike_type'])['rental_cost'].mean()
```

```python
# Which city had the highest average rating?
df.groupby('city')['rating'].mean().sort_values(ascending = False).head(5)
```

```python
# What is the total revenue per city on weekends only?
df[df['is_weekend']].groupby('city')['rental_cost'].sum()
```

```python
# get count, mean , and std of rental_cost grouped by mebership type.
df.groupby('membership_type')['rental_cost'].agg(['count','mean','std'])
```

```python
# What is the average rental duration for each gender and weather condition?
df.groupby(['gender','weather'])['duration_hours'].mean()
```

```python
#  How many 5-star ratings did each bike type receive?
df[df['rating']==5].groupby('bike_type').size()
```

```python
#  Rank cities by total number of rentals (highest first).
df['city'].value_counts()
```

```python
# What is the percentage share of each payment method per city?
df.groupby(['city','payment_method']).size().groupby(level=0).apply(lambda x: x/x.sum()*100).round(2)
```

```python
# Find the city-membership combination with the highest average rental cost.
avg_cost = df.groupby(['city','membership_type'])['rental_cost'].mean()
avg_cost.idxmax()
```
